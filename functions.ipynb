```python
# CARD DECK

def build_deck():
    suits = ["Hearts", "Spade", "Clubs", "Diamonds"]
    cards_numbers = ["A", "K", "Q", "J", "10", "9", "8", "7", "6", "5", "4", "3", "2"]

    return[(value, suit) for value in cards_numbers for suit in suits]


#card_deck = build_deck ()
#card_deck
#len(card_deck)
```

```python
#PLAYER COUNT

def counting_players():
    try:
        num_of_players = int(input("How many players are playing today?"))
    except ValueError:
        print("It needs to be a number")
        counting_players()
#number of players

    def create_player(num):
        return [{"name": "", "location": "" , "cards": "", "blind": "", "money": 1000} for i in range(0,num)]
        
    has_decided= False
    while not has_decided:
        if num_of_players > 1 and num_of_players < 7:
    # put the player creating logic DICTIONARY! 
            return create_player(num_of_players)
        else:
            try:
                num_of_players = int(input("How many players are playing? it has to be between 1-6"))
            except ValueError:
                print("It needs to be a number")
```
