## CREATING A LIST OF OPTIONS FOR THE PLAYER

```python
def player_options(player,pot):
    option_list = ["fold"]
    if sum(pot.values()) == 0:
        option_list.append("check")
    elif max(pot.values()) > pot[player["name"]]:
        option_list.append("call")

    option_list.append("raise")

    return option_list


```

## CHECK IF PLAYER HAS MONEY ( USED IN BET FUNC)

```python
def has_money(player,player_list, ammount):
    for i in player_list:
        if i["name"] == player:
            if ammount > i["money"]:
                return False
    return True
```

## BET FUNCTION WHERE MANEY GOES FROM PLAYER TO POT

```python
def bet(player,pot,player_list,ammount):
    if has_money(player,player_list,ammount):
        for i in player_list:
            if i["name"] == player["name"]:
                i["money"] = i["money"] - ammount
                pot[player["name"]] = pot[player["name"]] + ammount
                return True

    else:
        return False
    
```

## MAIN LOGIC WHERE THE PLAYER CAN DO LOGICAL DECISIONS

```python
def player_decision(player,pot,player_list):

    options = player_options(player,pot)
    player_choice = input(player["name"] +" what option you choose? {}".format(options))

    if player_choice.lower() in options:

        if player_choice.lower() == "fold":
            pot["total"] = pot["total"] + pot[player["name"]]
            pot.pop(player["name"])
            return True, "fold"
        #CALL
        elif player_choice.lower() == "call":
            value_to_pay = max(pot.values()) - pot[player["name"]]
            player_asnwer = input(player["name"] + " you will call {}€! write yes to procede".format(value_to_pay))

            if player_asnwer.lower().startswith("y"):

                if bet(player,pot,player_list,max(pot.values()) - pot[player["name"]]):
                    return True, "call"

                else:
                    input(player["name"] + " you had no money for that call, we will fold for you :)")
                    pot["total"] = pot["total"] + pot[player["name"]]
                    pot.pop(player["name"])
                    return True, "fold"
        #RAISE   
        elif player_choice.lower() == "raise":
            player_asnwer = input(player["name"] + " how much you want to raise 100 from ? write yes to accept")

            if player_asnwer.lower().startswith("y") and bet(player,pot,player_list,100):
                return True, "raise"

            else:
                input(player["name"] + " you are poor we will fold for you :)")
                pot["total"] = pot["total"] + pot[player["name"]]
                pot.pop(player["name"])
                return True, "fold"


        elif player_choice.lower() == "fold":

                if player_choice.lower() == "fold":
                    pot["total"] = pot["total"] + pot[player["name"]]
                    pot.pop((player["name"]))
                    return True, "fold"
    else:
        player_decision(player,pot,player_list)
```

```python
def is_pair(value_list):
    if max(value_list) > 1:
            return True

def is_trio(value_list):
      if max(value_list) > 2:
            return True


def is_flush(suit_list):
      if max(suit_list) >= 5:
            return True
```

```python
def is_round_over(pot):
    if pot <= 2:
        return True
    else:
        False
```

```python
def pay_pot(table,player_list,pot):
    hand_power_list = rank_hands(table,player_list)
    max_power = 0
    current_winner = ""
    for player, power in hand_power_list:
        if power > max_power:
            current_winner = player
    for player in player_list:
        if player["name"] == current_winner:
            player["money"] = player["money"] + pot["total"] + pot[player["name"]]
```
