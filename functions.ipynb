## CREATING A LIST OF OPTIONS FOR THE PLAYER

```python
def player_options(player,pot):
    option_list = ["fold"]
    if sum(pot.values()) == 0:
        option_list.append("check")
    elif max(pot.values()) > pot.get(player):
        option_list.append("call")

    option_list.append("raise")

    return option_list


```

## CHECK IF PLAYER HAS MONEY ( USED IN BET FUNC)

```python
def has_money(player,player_list, ammount):
    for i in player_list:
        if i["name"] == player:
            if ammount > i["money"]:
                return False
    return True
```

## BET FUNCTION WHERE MANEY GOES FROM PLAYER TO POT

```python
def bet(player,pot,player_list,ammount):
    if has_money(player,player_list,ammount):
        for i in player_list:
            if i["name"] == player["name"]:
                i["money"] = i["money"] - ammount
                pot[player] = pot[player] + ammount
                return True

    else:
        return False
    
```
