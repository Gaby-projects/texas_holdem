## DECK BUILDER

```python
def build_deck():
    suits = ["Hearts", "Spade", "Clubs", "Diamonds"]
    cards_numbers = ["A", "K", "Q", "J", "10", "9", "8", "7", "6", "5", "4", "3", "2"]

    return[(value, suit) for value in cards_numbers for suit in suits]
```

```python
def deal_cards(player_list,deck):
        counter = 0
        
        for index, card in enumerate(deck[:len(player_list)*2]):
                if index < len(player_list):
                        player_list[index]["cards"] = card
                elif index >= len(player_list):
                        index = counter
                        counter += 1
                        player_list[index]["cards"] = [player_list[index].get("cards"), card]
        

        remove_card(deck, 2*len(player_list))
```

## CREATING A LIST OF OPTIONS FOR THE PLAYER

```python
def player_options(player,pot):
    option_list = ["fold"]
    if sum(pot.values()) == 0:
        option_list.append("check")
    elif max(pot.values()) > pot[player["name"]]:
        option_list.append("call")

    option_list.append("raise")

    return option_list


```

## CHECK IF PLAYER HAS MONEY ( USED IN BET FUNC)

```python
def has_money(player,player_list, ammount):
    for i in player_list:
        if i["name"] == player:
            if ammount > i["money"]:
                return False
    return True
```

## BET FUNCTION WHERE MANEY GOES FROM PLAYER TO POT

```python
def bet(player,pot,player_list,ammount):
    if has_money(player,player_list,ammount):
        for i in player_list:
            if i["name"] == player["name"]:
                i["money"] = i["money"] - ammount
                pot[player["name"]] = pot[player["name"]] + ammount
                return True

    else:
        return False
    
```

## MAIN LOGIC WHERE THE PLAYER CAN DO LOGICAL DECISIONS

```python
def player_decision(player,pot,player_list):

    if len(pot) == 2:
        print("inside player decision {}".format(pot))
        return False, "win"

    options = player_options(player,pot)
    player_choice = input(player["name"] +" what option you choose? {}".format(options))

    if player_choice.lower() in options:

        if player_choice.lower() == "fold":
            pot["total"] = pot["total"] + pot[player["name"]]
            del pot[player["name"]]
            return True, "fold"
        #CALL
        elif player_choice.lower() == "call":
            value_to_pay = max(pot.values()) - pot[player["name"]]
            player_asnwer = input(player["name"] + " you will call {}€! write yes to procede".format(value_to_pay))

            if player_asnwer.lower().startswith("y"):

                if bet(player,pot,player_list,max(pot.values()) - pot[player["name"]]):
                    return True, "call"

                else:
                    input(player["name"] + " you had no money for that call, we will fold for you :)")
                    pot["total"] = pot["total"] + pot[player["name"]]
                    pot.pop(player["name"])
                    return True, "fold"
        #RAISE   
        elif player_choice.lower() == "raise":
            player_asnwer = input(player["name"] + " how much you want to raise 100 from ? write yes to accept")

            if player_asnwer.lower().startswith("y") and bet(player,pot,player_list,100):
                return True, "raise"

            else:
                input(player["name"] + " you are poor we will fold for you :)")
                pot["total"] = pot["total"] + pot[player["name"]]
                pot.pop(player["name"])
                return True, "fold"


        elif player_choice.lower() == "fold":

                if player_choice.lower() == "fold":
                    pot["total"] = pot["total"] + pot[player["name"]]
                    pot.pop((player["name"]))
                    return True, "fold"
    else:
        player_decision(player,pot,player_list)
```

## RANKING THE CARDS 

```python
## cards ranking

from collections import Counter

def rank_hands(table, player_list,pot):
    scores = [] # where we store ("Player name", int(score)) 
    for player in players_in_round(player_list,pot):        
        
        player["cards"] = player["cards"] + table
        card_counter_values = Counter(x for (x,y) in player["cards"])
        card_counter_suits = Counter(y for (x,y) in player["cards"])
        values_list = card_counter_values.values()
        suits_list = card_counter_suits.values()

        if is_flush(suits_list):
            scores.append((player["name"],3))
        elif is_trio(values_list):
            scores.append((player["name"],2))
        elif is_pair(values_list):
            scores.append((player["name"],1))

    return scores
        


```

```python
def is_pair(value_list):
    if max(value_list) > 1:
            return True

def is_trio(value_list):
      if max(value_list) > 2:
            return True


def is_flush(suit_list):
      if max(suit_list) >= 5:
            return True
```

```python
def is_round_over(pot):
    if len(pot) <= 2:
        return True
    else:
        return False
```

```python
def pay_pot(table,player_list,pot):
    hand_power_list = rank_hands(table,player_list,pot)
    max_power = -200
    current_winner = ""
    players_live = players_in_round(player_list, pot)
    if players_live == 1:
        current_winner = players_live[0]
    else:
        for player, power in hand_power_list:
            if power > max_power:
                current_winner = player
    total_payout = 0
    for i in pot:
        total_payout += i
    
    for player in player_list:
        if player["name"] == current_winner:
            player["money"] = player["money"] + total_payout
            input(player["name"] + "just won the round!!!")
```
